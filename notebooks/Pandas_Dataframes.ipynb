Creating dataframes

The whole table like structure is called dataframe in pandas and each col is called series and similarly every row is also a series

```python
# using lists
students_data = [
    [100,80,10],
    [90,70,7],
    [90,80,10],
    [80,50,2]
]
pd.DataFrame(students_data,columns =['iq','marks','package'])
```

```python
# using dicts
student_dict={
    'name':['A','B','C','D'],
    'iq':[100,90,90,80],
    'marks':[80,70,80,50],
    'package':[10,7,10,2]
}
students = pd.DataFrame(student_dict)
students.set_index('name',inplace=True)
students
```
